```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Data Loading

The raw dataset can be found in `data/raw/results.csv`. The remote dataset can be found [here](https://media.githubusercontent.com/media/StackExchange/Survey/refs/heads/main/packages/archive/2025/results.csv).


```python
url = "../data/raw/results.csv"

df = pd.read_csv(url, low_memory=False)

# Using low_memory=False to get more information on columns with mixed types

df.head()
```

#### 1. `MainBranch`

\- There were six unique categories in this column. With non-developer entries dropped, there are only 37,467 rows left

\- No null values in this column

```python
df["MainBranch"].hist(bins=20, edgecolor="black")
plt.title("Main Branch Distribution")
plt.xlabel("MainBranch")
plt.ylabel("Frequency")
plt.xticks(rotation=90)
plt.show()

df["MainBranch"].describe()
df["MainBranch"].value_counts()
df["MainBranch"].isna().sum()
```

#### 2. `Age`

\- There are seven unique age categories

\- Most popular age range is `25 - 34` years old with 13,919 entries, accounting for ~37% of total respondents. `35 - 44` years old and `18 - 24` years old come in 2nd and 3rd most frequent

\- No null values

```python
df["Age"].hist(bins=20, edgecolor="black")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.xticks(rotation=90)
plt.show()

df["Age"].describe()
df["Age"].value_counts()
df["Age"].isna().sum()
(df["Age"].isna().sum() / df.shape[0]) * 100
```
